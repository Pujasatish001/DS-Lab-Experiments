```python

# EXPERIMENT 9: Python Data Types, Variables, Expressions, Objects and Functions: 21-11-2026

# 1. Declare different data types
age = 21
weight = 65.5
name = "Arun"
is_patient = True
address = None
number = 3 + 4j

print("1. DATA TYPES")
print(age, type(age))
print(weight, type(weight))
print(name, type(name))
print(is_patient, type(is_patient))
print(address, type(address))
print(number, type(number))


# 2. Arithmetic, relational, logical and type conversion expressions
print("\n2. EXPRESSIONS")

a = 10
b = 5

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)

print("Relational:", a > b)
print("Logical:", a > 0 and b > 0)

x = "100"
print("Type Conversion:", int(x) + 50)


# 3. Functions to calculate BMI and health status

def calculate_bmi(weight, height):
    return weight / (height * height)


def bmi_category(bmi):
    if bmi < 18.5:
        return "Underweight"
    elif bmi < 25:
        return "Normal"
    elif bmi < 30:
        return "Overweight"
    else:
        return "Obese"


def bp_status(systolic, diastolic):
    if systolic < 120 and diastolic < 80:
        return "Normal"
    elif systolic < 130 and diastolic < 80:
        return "Elevated"
    else:
        return "High Blood Pressure"


# 4. Default arguments, *args, **kwargs and lambda

def greet(name="Patient"):
    print("Hello", name)


def total(*args):
    return sum(args)


def patient_details(**kwargs):
    for key, value in kwargs.items():
        print(key, ":", value)


square = lambda n: n * n

print("\n3. FUNCTIONS")
greet("Arun")
print("Total:", total(10, 20, 30))
patient_details(Name="Arun", Age=21, City="Chennai")
print("Square:", square(5))


# 5. Create Patient class with attributes and methods

class Patient:

    clinic_name = "ABC Health Clinic"

    def __init__(self, name, age, weight, height, systolic, diastolic):
        self.name = name
        self.age = age
        self.weight = weight
        self.height = height
        self.systolic = systolic
        self.diastolic = diastolic

    def get_bmi(self):
        return calculate_bmi(self.weight, self.height)

    def get_report(self):
        bmi = self.get_bmi()

        print("\n----- PATIENT HEALTH REPORT -----")
        print("Clinic:", Patient.clinic_name)
        print("Name:", self.name)
        print("Age:", self.age)
        print("BMI:", round(bmi, 2))
        print("BMI Category:", bmi_category(bmi))
        print("Blood Pressure:", bp_status(self.systolic, self.diastolic))

    def __str__(self):
        return self.name + " - " + str(self.age) + " years"


# Create patient objects
p1 = Patient("Arun", 21, 65, 1.7, 118, 76)
p2 = Patient("Priya", 25, 80, 1.6, 140, 90)

print("\n4. PATIENT OBJECTS")
print(p1)
print(p2)

p1.get_report()
p2.get_report()
```

```python
#Experiment 10:21-11-2026
import re
import array
import numpy as np

# 1. STRING OPERATIONS

review = "  This Product is Very Good and Useful  "

print("1. STRING OPERATIONS")
print("Original:", review)
print("Strip:", review.strip())
print("Lower:", review.lower())
print("Upper:", review.upper())
print("Slicing:", review[2:15])
print("Find:", review.find("Good"))
print("Count:", review.count("o"))
print("Replace:", review.replace("Good", "Excellent"))
print("Split:", review.split())
print("Join:", "-".join(review.split()))
print("Format:", "Customer Review: {}".format(review.strip()))


# 2. SENTIMENT SCORE

reviews = [
    "This product is good and excellent",
    "The product is bad and poor",
    "The product is useful and amazing"
]

positive = ["good", "excellent", "useful", "amazing", "great"]
negative = ["bad", "poor", "worst", "waste", "slow"]

print("\n2. SENTIMENT ANALYSIS")

for r in reviews:
    score = 0
    words = r.lower().split()

    for word in words:
        if word in positive:
            score += 1
        elif word in negative:
            score -= 1

    print("Review:", r)
    print("Sentiment Score:", score)


# 3. EXTRACT EMAIL ADDRESSES USING REGEX

text = "Contact us at abc@gmail.com or support@yahoo.com"

emails = re.findall(r'[\w.-]+@[\w.-]+\.\w+', text)

print("\n3. EMAIL EXTRACTION")
print("Email Addresses:", emails)


# 4. ARRAY MODULE OPERATIONS

print("\n4. ARRAY OPERATIONS")

arr = array.array('i', [10, 20, 30, 40, 50])

print("Original Array:", arr.tolist())

arr.append(60)
print("After Append:", arr.tolist())

arr.insert(1, 15)
print("After Insert:", arr.tolist())

arr.remove(30)
print("After Remove:", arr.tolist())

print("Array Element:", arr[0])
print("Array Slice:", arr[1:4].tolist())


# 5. NUMPY 2-D ARRAY OPERATIONS

print("\n5. NUMPY ARRAY OPERATIONS")

data = np.array([
    [25, 27, 26, 28, 29, 30, 31],
    [24, 26, 25, 27, 28, 29, 30],
    [26, 28, 27, 29, 30, 31, 32]
])

print("Original 2-D Array:")
print(data)

# Indexing
print("Indexing:", data[0, 2])

# Slicing
print("Slicing:")
print(data[:, 1:4])

# Boolean Filtering
print("Temperatures above 28:")
print(data[data > 28])

# Aggregation
print("Maximum:", np.max(data))
print("Minimum:", np.min(data))
print("Average:", np.mean(data))
print("Sum:", np.sum(data))

# Reshaping
new_data = data.reshape(7, 3)
print("Reshaped Array:")
print(new_data)

# Vectorised Arithmetic
print("Temperature after adding 2:")
print(data + 2)

print("Temperature after subtracting 2:")
print(data - 2)
```

```python

# EXPERIMENT 11: List and Tuple Operations: 21-11-2026
# Pharmacy Inventory Management System

# 1. Create a tuple of product records
products = (
    (101, "Paracetamol", 20),
    (102, "Vitamin C", 50),
    (103, "Cough Syrup", 80),
    (104, "Pain Relief", 100)
)

print("1. PRODUCT CATALOGUE")
print(products)

# Indexing
print("First Product:", products[0])

# Slicing
print("First Two Products:", products[:2])

# Unpacking
code, name, price = products[0]
print("Unpacked:", code, name, price)

# Count and Index
print("Count:", products.count(products[0]))
print("Index:", products.index(products[1]))


# 2. Tuples are immutable
print("\n2. TUPLE IMMUTABILITY")

try:
    products[0] = (105, "Tablet", 30)
except TypeError:
    print("Tuple cannot be modified")


# 3. List methods
print("\n3. LIST OPERATIONS")

stock = [50, 30, 20, 40]

print("Original Stock:", stock)

stock.append(60)
print("Append:", stock)

stock.extend([70, 80])
print("Extend:", stock)

stock.insert(1, 25)
print("Insert:", stock)

stock.remove(20)
print("Remove:", stock)

stock.pop()
print("Pop:", stock)

stock.sort()
print("Sort:", stock)

stock.reverse()
print("Reverse:", stock)

new_stock = stock.copy()
print("Copy:", new_stock)

temp = stock.copy()
temp.clear()
print("Clear:", temp)


# 4. List comprehension
print("\n4. STOCK UPDATE")

stock = [50, 30, 10, 40]

# Increase stock by 5
updated_stock = [x + 5 for x in stock]

print("Original Stock:", stock)
print("Updated Stock:", updated_stock)

# Create inventory records
inventory = [
    (products[i][0], products[i][1], products[i][2], updated_stock[i])
    for i in range(len(products))
]

print("Inventory Records:")
for item in inventory:
    print(item)


# 5. Reorder items and calculate total stock value
print("\n5. INVENTORY REPORT")

# Items with stock less than 20
reorder = [inventory[i][1] for i in range(len(inventory))
           if inventory[i][3] < 20]

print("Items to Reorder:", reorder)

# Calculate total stock value
total = sum(item[2] * item[3] for item in inventory)

print("Total Stock Value:", total)


# Nested sales list
sales = [
    ["Paracetamol", [10, 15, 20]],
    ["Vitamin C", [5, 10, 15]],
    ["Cough Syrup", [8, 12, 10]],
    ["Pain Relief", [4, 6, 8]]
]

print("\nSales Summary")

for item in sales:
    print("Product:", item[0])
    print("Total Sales:", sum(item[1]))
```

```python
#Experiment 12: 23-11-2026
from collections import defaultdict, Counter

# 1. SET OPERATIONS

python = {"Arun", "Priya", "Rahul", "John"}
ml = {"Priya", "Rahul", "Kavi", "David"}
bigdata = {"Arun", "Kavi", "David", "Ravi"}

print("1. SET OPERATIONS")

print("Union:", python | ml)
print("Intersection:", python & ml)
print("Difference:", python - ml)
print("Symmetric Difference:", python ^ ml)


# 2. SUBSET, DISJOINT, ADD AND DISCARD

print("\n2. SET RELATIONS")

print("Subset:", {"Priya", "Rahul"}.issubset(python))
print("Disjoint:", python.isdisjoint(bigdata))

python.add("Meena")
print("After Add:", python)

python.discard("John")
print("After Discard:", python)


# 3. NESTED DICTIONARY OPERATIONS

print("\n3. DICTIONARY OPERATIONS")

marks = {
    "Arun": {"Python": 85, "ML": 90},
    "Priya": {"Python": 92, "ML": 88},
    "Rahul": {"Python": 78, "ML": 82}
}

print("Original Dictionary:", marks)

# Access
print("Arun's Python Mark:", marks["Arun"]["Python"])

# Insert
marks["Kavi"] = {"Python": 80, "ML": 86}
print("After Insert:", marks)

# Update
marks["Arun"]["Python"] = 95
print("After Update:", marks)

# Delete
del marks["Rahul"]
print("After Delete:", marks)


# 4. DICTIONARY COMPREHENSION, GET, DEFAULTDICT AND SORTING

print("\n4. DICTIONARY METHODS")

# Dictionary comprehension
total_marks = {
    name: sum(subjects.values())
    for name, subjects in marks.items()
}

print("Total Marks:", total_marks)

# get() method
print("Arun's Marks:", marks.get("Arun"))
print("Unknown Student:", marks.get("John", "Not Found"))

# defaultdict
student_count = defaultdict(int)

student_count["Python"] += 1
student_count["Python"] += 1
student_count["ML"] += 1

print("Course Count:", dict(student_count))

# Sorting by value
sorted_marks = sorted(total_marks.items(),
                      key=lambda x: x[1],
                      reverse=True)

print("Sorted Marks:", sorted_marks)


# 5. WORD FREQUENCY USING COUNTER

print("\n5. WORD FREQUENCY")

feedback = """
Python is easy to learn.
Python is useful and easy.
Machine learning is interesting.
Python is interesting.
"""

words = feedback.lower().split()

frequency = Counter(words)

print("Word Frequency:")
print(frequency)

print("Most Common Words:")
print(frequency.most_common(3))
```
