```python
#Experiment 17: 24-11-2026
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import matplotlib.dates as mdates

# Create 24 months of date data
dates = pd.date_range(start='2024-01-01', periods=24, freq='MS')

# Monthly sales data
electronics = [120, 125, 130, 128, 135, 140, 145, 142,
               150, 155, 160, 180, 150, 155, 160, 158,
               165, 170, 175, 172, 180, 185, 190, 210]

clothing = [80, 85, 90, 88, 95, 100, 105, 110,
            100, 105, 120, 140, 90, 95, 100, 105,
            110, 115, 120, 125, 130, 135, 145, 160]

groceries = [100, 102, 105, 108, 110, 112, 115, 118,
             120, 122, 125, 130, 105, 108, 110, 112,
             115, 118, 120, 123, 125, 128, 132, 135]

# Create DataFrame
df = pd.DataFrame({
    'Electronics': electronics,
    'Clothing': clothing,
    'Groceries': groceries
}, index=dates)

# Calculate 3-month moving average
df['Electronics_MA'] = df['Electronics'].rolling(3).mean()

# Calculate month-on-month growth
growth = df[['Electronics', 'Clothing', 'Groceries']].pct_change() * 100

# Create figure with two subplots
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 10))

# Plot sales lines
ax1.plot(dates, electronics, color='blue',
         linestyle='-', marker='o', label='Electronics')

ax1.plot(dates, clothing, color='red',
         linestyle='--', marker='s', label='Clothing')

ax1.plot(dates, groceries, color='green',
         linestyle='-.', marker='^', label='Groceries')

# Plot linear trend line for Electronics
x = np.arange(len(dates))
slope, intercept = np.polyfit(x, electronics, 1)
trend = slope * x + intercept

ax1.plot(dates, trend, color='purple',
         linestyle=':', linewidth=2, label='Linear Trend')

# Plot 3-month moving average
ax1.plot(dates, df['Electronics_MA'],
         color='orange', linewidth=2,
         label='3-Month Moving Average')

# Add title and labels
ax1.set_title('Monthly Sales Trends (2024-2025)',
              fontsize=16, fontweight='bold')

ax1.set_xlabel('Month')
ax1.set_ylabel('Sales (in thousands)')

# Set axis limits
ax1.set_ylim(50, 250)

# Format date ticks
ax1.xaxis.set_major_locator(mdates.MonthLocator(interval=2))
ax1.xaxis.set_major_formatter(mdates.DateFormatter('%b-%Y'))

# Rotate date labels
plt.setp(ax1.get_xticklabels(), rotation=45)

# Add grid and legend
ax1.grid(True, linestyle='--', alpha=0.6)
ax1.legend()

# Add annotation with arrow
ax1.annotate('Highest Electronics Sales',
             xy=(dates[23], electronics[23]),
             xytext=(dates[16], 230),
             arrowprops=dict(facecolor='black', arrowstyle='->'),
             fontsize=10)

# Add text label
ax1.text(dates[2], 220, 'Retail Sales Analysis',
         fontsize=11, color='brown')

# Plot month-on-month growth
ax2.plot(dates, growth['Electronics'],
         marker='o', label='Electronics Growth')

ax2.plot(dates, growth['Clothing'],
         marker='s', label='Clothing Growth')

ax2.plot(dates, growth['Groceries'],
         marker='^', label='Groceries Growth')

ax2.set_title('Month-on-Month Sales Growth',
              fontsize=14)

ax2.set_xlabel('Month')
ax2.set_ylabel('Growth (%)')

ax2.xaxis.set_major_locator(mdates.MonthLocator(interval=2))
ax2.xaxis.set_major_formatter(mdates.DateFormatter('%b-%Y'))

plt.setp(ax2.get_xticklabels(), rotation=45)

ax2.grid(True, linestyle='--', alpha=0.6)
ax2.legend()

# Adjust layout
plt.tight_layout()

# Save figure
plt.savefig('sales_trends.png', dpi=300)

# Display graph
plt.show()

print("Sales trend analysis:")
print("Electronics shows an overall increasing trend.")
print("Clothing shows seasonal variations in sales.")
print("Groceries shows steady growth over time.")
print("\nMonth-on-month growth:")
print(growth.round(2))
```

```python
#Experiment 18:25-11-2026

import matplotlib.pyplot as plt
import numpy as np

# Hospital department data
departments = ['Cardiology', 'Neurology', 'Orthopedics', 'Pediatrics']

# Budget allocation
budget = [30, 25, 20, 25]

# Patient counts for two years
patients_2024 = [450, 350, 400, 300]
patients_2025 = [500, 400, 450, 350]

# Patient age distribution
ages = [22, 25, 30, 35, 40, 45, 50, 55,
        60, 65, 70, 32, 28, 38, 42, 48,
        52, 58, 62, 68, 72, 75, 80, 45,
        35, 29, 33, 50, 55, 60]

# Waiting times in minutes for each department
cardiology = [20, 25, 30, 35, 40, 45, 50]
neurology = [15, 20, 25, 30, 35, 40, 45]
orthopedics = [25, 30, 35, 40, 45, 50, 55]
pediatrics = [10, 15, 20, 25, 30, 35, 40]

# Create dashboard
fig, ax = plt.subplots(2, 2, figsize=(14, 10))

fig.suptitle('Hospital Operations Dashboard',
             fontsize=18, fontweight='bold')

# 1. Pie Chart - Budget Distribution
explode = (0.1, 0, 0, 0)

ax[0, 0].pie(budget,
             labels=departments,
             autopct='%1.1f%%',
             explode=explode,
             startangle=90)

ax[0, 0].set_title('Department Budget Distribution')

# 2. Grouped Bar Chart - Patient Comparison
x = np.arange(len(departments))
width = 0.35

bars1 = ax[0, 1].bar(x - width/2, patients_2024,
                     width, label='2024', color='skyblue')

bars2 = ax[0, 1].bar(x + width/2, patients_2025,
                     width, label='2025', color='orange')

ax[0, 1].set_title('Patient Count Comparison')
ax[0, 1].set_xlabel('Departments')
ax[0, 1].set_ylabel('Number of Patients')

ax[0, 1].set_xticks(x)
ax[0, 1].set_xticklabels(departments, rotation=15)

ax[0, 1].legend()

ax[0, 1].bar_label(bars1)
ax[0, 1].bar_label(bars2)

# 3. Histogram - Patient Age Distribution
ax[1, 0].hist(ages, bins=8,
              color='lightgreen',
              edgecolor='black')

mean_age = np.mean(ages)

ax[1, 0].axvline(mean_age,
                 color='red',
                 linestyle='--',
                 linewidth=2,
                 label=f'Mean Age = {mean_age:.1f}')

ax[1, 0].set_title('Patient Age Distribution')
ax[1, 0].set_xlabel('Age')
ax[1, 0].set_ylabel('Number of Patients')
ax[1, 0].legend()

# 4. Box Plot - Waiting Time Comparison
waiting_times = [cardiology, neurology,
                 orthopedics, pediatrics]

ax[1, 1].boxplot(waiting_times,
                 tick_labels=departments)

ax[1, 1].set_title('Department-wise Waiting Times')
ax[1, 1].set_xlabel('Departments')
ax[1, 1].set_ylabel('Waiting Time (Minutes)')

ax[1, 1].tick_params(axis='x', rotation=15)

# Adjust dashboard layout
plt.tight_layout(rect=[0, 0, 1, 0.95])

# Save dashboard
plt.savefig('hospital_dashboard.png', dpi=300)

# Display dashboard
plt.show()

# Print chart justifications
print("\nCHART SELECTION JUSTIFICATION")

print("1. Pie Chart: Shows the percentage budget share "
      "of each department.")

print("2. Grouped Bar Chart: Compares patient counts "
      "between 2024 and 2025.")

print("3. Histogram: Shows the distribution of patient ages "
      "and marks the mean age.")

print("4. Box Plot: Compares the spread, median and "
      "variation of waiting times across departments.")

print("\nHospital dashboard created successfully!")
print("Mean patient age:", round(mean_age, 2))
```

```python
#Experiment 19: 25-11-2026
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

# Handle Basemap deprecation warnings
warnings.filterwarnings("ignore", category=DeprecationWarning)

# Import Basemap
try:
    from mpl_toolkits.basemap import Basemap
    basemap_available = True
except ImportError:
    basemap_available = False
    print("Basemap is not installed. Map cannot be displayed.")

# Create air pollution dataset
data = {
    'City': ['Delhi', 'Mumbai', 'Kolkata', 'Chennai',
             'Bangalore', 'Hyderabad', 'Pune', 'Ahmedabad',
             'Jaipur', 'Lucknow', 'Bhopal', 'Patna',
             'Surat', 'Nagpur', 'Kochi', 'Chandigarh'],

    'PM25': [95, 65, 75, 45, 35, 55, 40, 80,
             70, 85, 50, 90, 60, 55, 30, 38],

    'Vehicles': [900, 750, 700, 500, 450, 600, 480, 800,
                 650, 780, 520, 850, 620, 550, 350, 400],

    'Industries': [80, 90, 85, 50, 40, 65, 45, 95,
                   60, 75, 55, 70, 80, 50, 30, 35],

    'GreenCover': [15, 25, 20, 35, 45, 30, 40, 18,
                   22, 16, 38, 14, 28, 32, 50, 48],

    'Latitude': [28.61, 19.07, 22.57, 13.08,
                 12.97, 17.38, 18.52, 23.02,
                 26.91, 26.85, 23.25, 25.61,
                 21.17, 21.15, 9.93, 30.73],

    'Longitude': [77.20, 72.87, 88.36, 80.27,
                  77.59, 78.48, 73.86, 72.57,
                  75.78, 80.95, 77.41, 85.14,
                  72.83, 79.09, 76.27, 76.78]
}

df = pd.DataFrame(data)

# 1. Pearson Correlation Matrix
print("\nPEARSON CORRELATION MATRIX")
correlation = df[['PM25', 'Vehicles',
                  'Industries', 'GreenCover']].corr(method='pearson')

print(correlation.round(2))

# Describe correlation strength
print("\nCORRELATION ANALYSIS")

for column in ['Vehicles', 'Industries', 'GreenCover']:
    value = correlation.loc['PM25', column]

    if abs(value) >= 0.7:
        strength = "Strong"
    elif abs(value) >= 0.4:
        strength = "Moderate"
    else:
        strength = "Weak"

    direction = "Positive" if value > 0 else "Negative"

    print(column, ":", direction, strength,
          "correlation with PM2.5 =", round(value, 2))

# 2. Scatter Plot
plt.figure(figsize=(9, 6))

scatter = plt.scatter(
    df['Vehicles'],
    df['PM25'],
    c=df['GreenCover'],
    s=df['Industries'] * 5,
    cmap='viridis',
    alpha=0.8,
    edgecolors='black'
)

# Add city labels
for i in range(len(df)):
    plt.text(df['Vehicles'][i] + 5,
             df['PM25'][i] + 1,
             df['City'][i],
             fontsize=8)

# Fit a linear trend line
slope, intercept = np.polyfit(df['Vehicles'], df['PM25'], 1)

x = np.linspace(df['Vehicles'].min(),
                df['Vehicles'].max(), 100)

y = slope * x + intercept

plt.plot(x, y, color='red',
         linestyle='--', label='Fitted Line')

plt.colorbar(scatter, label='Green Cover')
plt.title('Vehicles vs PM2.5 Pollution')
plt.xlabel('Vehicle Count')
plt.ylabel('PM2.5 Level')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.savefig('pollution_scatter.png', dpi=300)
plt.show()

# 3. Correlation Heatmap
plt.figure(figsize=(8, 6))

sns.heatmap(correlation,
            annot=True,
            cmap='coolwarm',
            fmt='.2f',
            linewidths=0.5,
            vmin=-1,
            vmax=1)

plt.title('Air Pollution Correlation Heatmap')
plt.tight_layout()
plt.savefig('pollution_heatmap.png', dpi=300)
plt.show()

# 4. Geographic Map using Basemap
if basemap_available:

    plt.figure(figsize=(10, 10))

    try:
        m = Basemap(
            projection='merc',
            llcrnrlat=6,
            urcrnrlat=38,
            llcrnrlon=68,
            urcrnrlon=98,
            resolution='i'
        )

        m.drawcoastlines()
        m.drawcountries()
        m.drawmapboundary(fill_color='lightblue')
        m.fillcontinents(color='lightyellow',
                         lake_color='lightblue')

        # Draw parallels and meridians
        m.drawparallels(np.arange(0, 41, 5),
                        labels=[1, 0, 0, 0])

        m.drawmeridians(np.arange(65, 101, 5),
                        labels=[0, 0, 0, 1])

        # Convert longitude and latitude
        x, y = m(df['Longitude'].values,
                 df['Latitude'].values)

        # Plot cities using PM2.5 as colour
        points = m.scatter(
            x, y,
            c=df['PM25'],
            s=100,
            cmap='Reds',
            edgecolors='black',
            zorder=5
        )

        # Add city labels
        for i in range(len(df)):
            plt.text(x[i] + 30000,
                     y[i] + 30000,
                     df['City'][i],
                     fontsize=8)

        plt.colorbar(points, label='PM2.5 Level')

        plt.title('Air Pollution Levels Across Indian Cities')

        plt.tight_layout()
        plt.savefig('india_pollution_map.png', dpi=300)
        plt.show()

    except Exception as e:
        print("Error while creating Basemap:", e)

# 5. Final Result
print("\nAIR POLLUTION ANALYSIS")
print("Highest PM2.5 pollution:",
      df.loc[df['PM25'].idxmax(), 'City'])

print("Lowest PM2.5 pollution:",
      df.loc[df['PM25'].idxmin(), 'City'])

print("\nAll graphs saved successfully.")
```

```python
#Experiment: 25-11-2026


import networkx as nx
import matplotlib.pyplot as plt

# ==========================================
# 1. UNDIRECTED GRAPH - RESEARCH NETWORK
# ==========================================

G = nx.Graph()

# Add researchers
researchers = ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H']
G.add_nodes_from(researchers)

# Add collaboration edges
collaborations = [
    ('A', 'B'), ('A', 'C'), ('B', 'C'),
    ('B', 'D'), ('C', 'D'), ('D', 'E'),
    ('E', 'F'), ('F', 'G'), ('G', 'H'),
    ('E', 'G'), ('F', 'H')
]

G.add_edges_from(collaborations)

# Display nodes and edges
print("UNDIRECTED GRAPH - RESEARCH NETWORK")

print("Nodes:", G.nodes())
print("Edges:", G.edges())

# Calculate density and connectivity
print("Density:", round(nx.density(G), 2))
print("Is Connected:", nx.is_connected(G))

# Degree centrality
degree = nx.degree_centrality(G)

print("\nDegree Centrality:")
for node, value in degree.items():
    print(node, ":", round(value, 2))

# Betweenness centrality
betweenness = nx.betweenness_centrality(G)

print("\nBetweenness Centrality:")
for node, value in betweenness.items():
    print(node, ":", round(value, 2))

# Shortest path
path = nx.shortest_path(G, source='A', target='H')

print("\nShortest Path from A to H:", path)

# Community detection
communities = list(nx.community.greedy_modularity_communities(G))

print("\nCommunities:")
for i, community in enumerate(communities):
    print("Community", i + 1, ":", community)

# Assign community colours
colors = {}

for i, community in enumerate(communities):
    for node in community:
        colors[node] = i

node_colors = [colors[node] for node in G.nodes()]

# Node size based on degree centrality
node_sizes = [1000 + degree[node] * 3000 for node in G.nodes()]

# Highlight shortest path
path_edges = list(zip(path, path[1:]))

edge_colors = [
    'red' if edge in path_edges or edge[::-1] in path_edges
    else 'gray'
    for edge in G.edges()
]

# Draw undirected graph
plt.figure(figsize=(9, 7))

pos = nx.spring_layout(G, seed=42)

nx.draw(
    G, pos,
    with_labels=True,
    node_color=node_colors,
    node_size=node_sizes,
    edge_color=edge_colors,
    cmap=plt.cm.Set2,
    font_size=12,
    font_weight='bold',
    width=2
)

plt.title("Research Collaboration Network")
plt.savefig("research_network.png", dpi=300)
plt.show()


# ==========================================
# 2. DIRECTED WEIGHTED GRAPH - FLIGHT NETWORK
# ==========================================

D = nx.DiGraph()

# Add cities
cities = ['Delhi', 'Mumbai', 'Chennai',
          'Kolkata', 'Bangalore', 'Hyderabad']

D.add_nodes_from(cities)

# Add directed weighted flight routes
routes = [
    ('Delhi', 'Mumbai', 1150),
    ('Delhi', 'Chennai', 1760),
    ('Delhi', 'Kolkata', 1300),
    ('Mumbai', 'Bangalore', 840),
    ('Mumbai', 'Hyderabad', 620),
    ('Bangalore', 'Chennai', 290),
    ('Bangalore', 'Hyderabad', 500),
    ('Chennai', 'Kolkata', 1370),
    ('Hyderabad', 'Kolkata', 1180),
    ('Kolkata', 'Delhi', 1300)
]

D.add_weighted_edges_from(routes)

print("\nDIRECTED WEIGHTED GRAPH - FLIGHT NETWORK")

print("Nodes:", D.nodes())
print("Edges:", D.edges(data=True))

# In-degree and out-degree
print("\nIn-Degree:")
for node, value in D.in_degree():
    print(node, ":", value)

print("\nOut-Degree:")
for node, value in D.out_degree():
    print(node, ":", value)

# PageRank
pagerank = nx.pagerank(D, weight='weight')

print("\nPageRank:")
for node, value in pagerank.items():
    print(node, ":", round(value, 3))

# Dijkstra shortest route
shortest_route = nx.dijkstra_path(
    D,
    source='Delhi',
    target='Chennai',
    weight='weight'
)

distance = nx.dijkstra_path_length(
    D,
    source='Delhi',
    target='Chennai',
    weight='weight'
)

print("\nShortest Route from Delhi to Chennai:")
print(shortest_route)

print("Total Distance:", distance, "km")

# Node sizes based on PageRank
node_sizes = [
    1000 + pagerank[node] * 5000
    for node in D.nodes()
]

# Highlight shortest route
route_edges = list(zip(shortest_route, shortest_route[1:]))

edge_colors = [
    'red' if edge in route_edges else 'gray'
    for edge in D.edges()
]

# Draw directed graph
plt.figure(figsize=(10, 7))

pos = nx.spring_layout(D, seed=42)

nx.draw_networkx_nodes(
    D, pos,
    node_size=node_sizes,
    node_color=list(pagerank.values()),
    cmap=plt.cm.Blues
)

nx.draw_networkx_labels(
    D, pos,
    font_size=10,
    font_weight='bold'
)

nx.draw_networkx_edges(
    D, pos,
    edge_color=edge_colors,
    width=2,
    arrows=True,
    arrowsize=20,
    connectionstyle='arc3,rad=0.1'
)

# Display edge weights
edge_labels = nx.get_edge_attributes(D, 'weight')

nx.draw_networkx_edge_labels(
    D, pos,
    edge_labels=edge_labels,
    font_size=8
)

plt.title("Directed Weighted Flight Route Network")
plt.axis('off')
plt.savefig("flight_network.png", dpi=300)
plt.show()

print("\nBoth graphs created successfully!")
```
